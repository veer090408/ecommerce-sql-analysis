```python
import sqlite3
import pandas as pd

df_orders = pd.read_csv('olist_orders_dataset.csv')
df_items = pd.read_csv('olist_order_items_dataset.csv')
df_products = pd.read_csv('olist_products_dataset.csv')

conn = sqlite3.connect('ecommerce.db')


df_orders.to_sql('orders', conn, index=False, if_exists='replace')
df_items.to_sql('order_items', conn, index=False, if_exists='replace')
df_products.to_sql('products', conn, index=False, if_exists='replace')

print("Done")

sql = """ SELECT * from df_orders"""
```

```python
# Query 1: Counting total orders and revenue
query_summary = """
SELECT 
    COUNT(DISTINCT order_id) AS total_orders,
    ROUND(SUM(price), 2) AS total_revenue,
    ROUND(AVG(price), 2) AS avg_order_value
FROM order_items;
"""

df_summary = pd.read_sql_query(query_summary, conn)
df_summary
```

```python
# Query 2: Top 5 bestseller product according revenue
query_top_products = """
SELECT 
    p.product_category_name AS category,
    COUNT(i.order_id) AS total_items_sold,
    ROUND(SUM(i.price), 2) AS total_revenue
FROM order_items i
JOIN products p ON i.product_id = p.product_id
WHERE p.product_category_name IS NOT NULL
GROUP BY p.product_category_name
ORDER BY total_revenue DESC
LIMIT 5;
"""

df_top_products = pd.read_sql_query(query_top_products, conn)
df_top_products
```

```python
#Table visualisation
df_top_products.style.bar(subset=['total_revenue'], color="#9f41ce")
```

```python
# Query for monthly revenue
query_trend = """
SELECT 
    substr(shipping_limit_date, 1, 7) AS month,
    COUNT(order_id) AS total_orders,
    ROUND(SUM(price), 2) AS monthly_revenue
FROM order_items
GROUP BY month
ORDER BY month ASC;
"""

df_trend = pd.read_sql_query(query_trend, conn)

# first month
df_trend.head(10)
```
